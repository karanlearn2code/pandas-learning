```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Superstore.csv", encoding = "latin1")
```

```python
df.head()
```

```python
# 1.How many Orders/records are there ?
df.shape
```

```python
len(df)
```

```python
# 2. What is the total sales / revenue ? 
df['Sales'].sum()
```

```python
# 3.What is the total profit ?
df['Profit'].sum()
```

```python
# 4. Which category has the highest sales ?
df.groupby("Category")["Sales"].sum().sort_values(ascending = False)
```

```python
# 5. Which category has the highest profit ?
df.groupby("Category")["Profit"].sum().sort_values(ascending = False)
```

```python
# 6. What are the top 10 products by sales
df.groupby("Product Name")["Sales"].sum().sort_values(ascending = False ).head(10)
```

```python
# 7. what are the 10 products with lowest profit ?
df.groupby("Product Name")["Profit"].sum().sort_values().head(10)
```

```python
# 8. which cities generate the most sales ?
df.groupby("City")["Sales"].sum().sort_values(ascending = False).head(10)
```

```python
# 9. How to sales change by year?
df["Order Date"] = pd.to_datetime(df["Order Date"],dayfirst = True)
df["Year"] = df["Order Date"].dt.year
df.groupby("Year")["Sales"].sum()
df.groupby("Year")["Sales"].sum().sort_index()
```

```python
# 10.Which product category have negative profit ?
df[df["Profit"] < 0]
```

```python
# categories with negative total profit 
df.groupby("Category")["Profit"].sum().sort_values()
```

```python
# product with biggest losses 
df.groupby("Product Name")["Profit"].sum().sort_values().head(10)
```
